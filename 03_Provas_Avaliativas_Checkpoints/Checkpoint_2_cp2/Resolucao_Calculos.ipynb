Identificar os pontos de atrito (pain points), oportunidades de inovação e curvas emocionais antes e após a intervenção de design.

### Formulação Matemática Formal
$$\text{NPS} = \% \text{Promotores} - \% \text{Detratores}, \quad \text{CES} = \frac{\sum\text{Esforço Avaliado}}{N}$$


### Passo 1: Entrevistas em Profundidade e Pesquisa Etnográfica
Condução de 12 entrevistas contextuais no chão de fábrica mapeando dores reais: telas com excesso de texto denso, letras pequenas e falta de atalhos rápidos.


### Passo 2: Construção da Persona 'Carlos - Operador de Turno'
Carlos tem 42 anos, opera com luvas industriais e precisa registrar paradas em menos de 15 segundos sem desviar a atenção do maquinário pesado.


### Passo 3: Mapeamento dos Estágios da Jornada
Fases da jornada: Início do Turno -> Identificação de Alerta -> Diagnóstico da Causa -> Abertura de Chamado -> Retomada da Linha.


### Passo 4: Identificação do Vale de Frustração Emocional
O ponto mais crítico ocorre na fase de diagnóstico, onde o sistema antigo exigia navegar por 5 menus aninhados sem campo de busca.


### Passo 5: Curva Emocional Pós-Redesign
Com a nova interface de busca preditiva e cartões de ação rápida, a nota de esforço do usuário (CES) caiu de 5,8 para 1,9 em escala de 7 pontos.


```python
import numpy as np
import matplotlib.pyplot as plt

estagios = ['Início Turno', 'Alerta Linha', 'Diagnóstico', 'Abertura O.S.', 'Retomada']
humor_antes = [1, -1, -3, -2, 0]
humor_depois = [2, 1, 2, 3, 3]
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#24121e')
ax.set_facecolor('#160a12')
ax.plot(estagios, humor_antes, marker='o', color='#ef4444', lw=2, label='Jornada Antes (Pain Points)')
ax.plot(estagios, humor_depois, marker='s', color='#10b981', lw=2.5, label='Jornada Redesenhada (UCD)')
ax.axhline(0, color='#94a3b8', ls=':', alpha=0.5)
ax.set_title('Curva de Satisfação Emocional na Jornada do Operador', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_ylabel('Nível de Satisfação (-3 a +3)', color='#94a3b8', fontsize=9)
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#502842', alpha=0.5, ls=':')
ax.legend(facecolor='#361b2c', edgecolor='#502842', labelcolor='#ffffff', fontsize=8)
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=200, facecolor=fig.get_facecolor())
plt.close()
```
